```python

!pip -q install -U gradio tensorflow pillow
import numpy as np
import tensorflow as tf
import gradio as gr
from PIL import Image
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import (
    preprocess_input, decode_predictions
)
print("Loading CNN model...")
model = MobileNetV2(
    weights="imagenet",
    include_top=True
)
print("CNN model loaded successfully!")
def recognize_tool(image):
    if image is None:
        return "Please upload an image or take a photo.", {}
    image = image.convert("RGB").resize((224, 224))
    arr = np.array(image, dtype=np.float32)
    arr = np.expand_dims(arr, axis=0)
    arr = preprocess_input(arr)
    predictions = model.predict(arr, verbose=0)
    decoded = decode_predictions(predictions, top=5)[0]
    result = {}
    for _, label, confidence in decoded:
        result[label.replace("_", " ")] = float(confidence)
        top_label = decoded[0][1].replace("_", " ")
    top_confidence = float(decoded[0][2]) * 100
    message = (
        "Image Recognition Result\n\n"
        f"Predicted object: {top_label}\n"
        f"Confidence: {top_confidence:.2f}%\n\n"
        "Top 5 predictions are shown in the chart."
    )
    return message, result
demo = gr.Interface(
    fn=recognize_tool,
    inputs=gr.Image(
        sources=["upload", "webcam"],
        type="pil",
        label="Upload Tool Image or Open Camera"
    ),
    outputs=[
        gr.Textbox(label="Recognition Output", lines=6),
        gr.Label(label="Top 5 Predictions", num_top_classes=5)
    ],
    title="Smart Tool Recognition System",
    description=(
        "CNN image recognition using MobileNetV2. "
        "Upload an image or capture one with your camera."
    )
)
print("Starting application...")
demo.launch(share=True)
```
